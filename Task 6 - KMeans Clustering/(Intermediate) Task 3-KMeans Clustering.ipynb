# Delete non-numeric columns and target

```python
df.drop(['State', 'Churn'], axis='columns', inplace=True)
```

# Scaling

```python
df.dtypes
```

```python
df['International plan'] = df['International plan'].replace({'No': 0, 'Yes': 1})
df['Voice mail plan'] = df['Voice mail plan'].replace({'No': 0, 'Yes': 1})
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

# Use the Elbow Method to Determine the Best K Number

```python
inertia = []
K = range(1, 11)

for k in K:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(scaled_data)
    inertia.append(model.inertia_)
```

```python
plt.plot(K, inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.grid(True)
plt.show()
```

# Using K=3 based on Elbow

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(scaled_data)
```

```python
print(df[['Cluster'] + list(df.columns[:-1])].head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(
    x=df['Total day minutes'], 
    y=df['Total eve minutes'], 
    hue=df['Cluster'], 
    palette='Set1'
)
plt.title("Customer Clusters based on Call Minutes")
plt.xlabel("Total Day Minutes")
plt.ylabel("Total Evening Minutes")
plt.legend(title='Cluster')
plt.show()
```
